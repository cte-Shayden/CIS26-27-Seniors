# 10.2 The Anatomy Of a Computer

## Central Processing Unit (CPU)

The part of a computer that executes the machine instructions.

## Primary Storage

Electronic circiuts that can store data as long as they have electric power. e.g. the memory that you have installed on a system.

## Secondary Storage

Storage that persits without electricity. e.g. a hard disc.

## Hard Disk

A device that stores information on rotating platters with magnetic coating. (traditional hard drives) Newer solid state drives do not have platters, they store data on chips.

## Network

An interconnected system of computers and other devices.
