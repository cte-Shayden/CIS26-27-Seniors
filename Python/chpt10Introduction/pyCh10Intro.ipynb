# 10.8 Python Errors

## Syntax Error

compile (translation) error, that is caught before the program runs. in other words, it stops the program from being able to run

## Execption (runtime error)

The syntax is fine (it is legal), so the program runs, however, becuse of this exeption, program will stop and give you an error

## Logic Error

Type of run-time error, the program runs normally and givea a output as it is supposed to, the problem is that if you are not paying attention, the output is incorrrect. ex. print(2+2) Output: 28

logic errors can be difficult to spot, this is why you need to have an understanding of what the ouput should be.

# 10.9 Algorithm Design

## Algorithm

a recipe for a solution

## Algorithms should be:

### Unambigious 

Precice instructions at every step, no guesswork or personal opinion

example: "Find the best looking person."
"Tell me who the greatest game development studio is."

### Exectutable 

every step can actually be carried out in practice.

example "use next decades real interest rate."

### Terminating

The steps eventually come to an end.

Example:

"repeat forever"

### How do we describe an algorithm in pseudocode?

1. inputs and outputs - what goes in? what do we want to come out?

2. break it down - split the larger problem into smaller, executable subtasks (problems)

3. write pseudocode- order the steps so vaules exist before theyre used.

4. test it- work a problem by hand with sample vaulues.
